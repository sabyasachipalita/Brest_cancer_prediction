```python
data.drop ( 'id',axis=1,inplace=True)
```

```python
#check the multicolinearity
col_matrix=data.corr().abs()
mask=np.triu(np.ones_like(col_matrix,dtype=bool))
tri_df = col_matrix.mask(mask)
to_drop=[x for x in tri_df.columns if (tri_df[x]>0.92).any()]


to_drop




```

```python
df=data.drop(to_drop,axis=1)
print(df.shape[1])
```

```python
df.head()
```

```python
x=df.drop('diagnosis',axis=1)
y=df['diagnosis']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
#scaling data
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)

```

```python
x_train.shape
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# 1. Handle missing values
imputer = SimpleImputer(strategy='mean')  #
x_train = imputer.fit_transform(x_train)
x_test = imputer.transform(x_test)

# 2. Scale the features
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
# now we apply ml model
from sklearn.linear_model import LogisticRegression

l_reg=LogisticRegression()
l_reg.fit(x_train,y_train)
```

```python
y_pred=l_reg.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
print('training_accuracy',accuracy_score(y_train,l_reg.predict(x_train)))
l_reg_acc=accuracy_score(y_test,l_reg.predict(x_test))

print(' testing_accuracy ',l_reg_acc)
y_pred=l_reg.predict(x_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
